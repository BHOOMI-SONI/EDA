# Exploratory Data Analysis — Titanic Dataset

**Goal:** Explore passenger data from the RMS Titanic to understand what factors were associated with survival.

**Sections:** Load data → Basic checks → Missing values → Univariate analysis → Survival by category → Correlation → Conclusions

## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)
```

## 2. Load the dataset
We load Titanic directly from seaborn's built-in datasets (no file upload needed).

```python
df = sns.load_dataset('titanic')
df.head()
```

## 3. Basic structure

```python
print('Shape:', df.shape)
df.info()
```

```python
df.describe(include='all')
```

## 4. Missing values

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing = missing[missing > 0]
missing
```

```python
sns.barplot(x=missing.values, y=missing.index, hue=missing.index, legend=False, palette='mako')
plt.title('Missing values per column')
plt.xlabel('Count')
plt.show()
```

## 5. Overall survival rate

```python
survival_rate = df['survived'].mean() * 100
print(f'Overall survival rate: {survival_rate:.1f}%')

sns.countplot(x='survived', data=df, hue='survived', legend=False, palette='Set2')
plt.xticks([0, 1], ['Did not survive', 'Survived'])
plt.title('Survival counts')
plt.show()
```

## 6. Survival by passenger class

```python
sns.barplot(x='pclass', y='survived', data=df, hue='pclass', legend=False, palette='Blues')
plt.title('Survival rate by passenger class')
plt.ylabel('Survival rate')
plt.show()
```

## 7. Survival by sex

```python
sns.barplot(x='sex', y='survived', data=df, hue='sex', legend=False, palette='Set1')
plt.title('Survival rate by sex')
plt.ylabel('Survival rate')
plt.show()
```

## 8. Survival by sex and class combined

```python
sns.catplot(x='pclass', y='survived', hue='sex', data=df, kind='bar', palette='coolwarm')
plt.title('Survival rate by class and sex')
plt.show()
```

## 9. Age distribution and survival

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.histplot(data=df, x='age', hue='survived', bins=30, kde=True, multiple='stack', palette='Set2')
plt.title('Age distribution by survival')
plt.show()
```

## 10. Fare distribution and survival

```python
sns.boxplot(x='survived', y='fare', data=df, hue='survived', legend=False, palette='Set3')
plt.ylim(0, 300)
plt.title('Fare by survival')
plt.show()
```

## 11. Family size feature

```python
df['family_size'] = df['sibsp'] + df['parch'] + 1
sns.barplot(x='family_size', y='survived', data=df, hue='family_size', legend=False, palette='viridis')
plt.title('Survival rate by family size')
plt.show()
```

## 12. Correlation heatmap (numeric features)

```python
num_cols = ['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare', 'family_size']
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation matrix')
plt.show()
```

## 13. Key findings

- Overall survival rate was roughly 38%.
- **Sex** was the strongest single predictor: women survived at a much higher rate than men.
- **Passenger class** mattered a lot: 1st class passengers had a much higher survival rate than 3rd class, consistent with 'women and children first' being applied more effectively for higher classes.
- **Age** had missing values (~20%) and very young children had somewhat better survival odds.
- **Fare** correlates with class (higher fare → higher class → higher survival), so fare and class carry overlapping information.
- Passengers traveling completely alone (family_size = 1) had lower survival rates than those with a small family (2-4).
